```python
import os
import shutil
import random
from glob import glob
```

```python
SOURCE_DIR = "../data/MIMII/fan"
DEST_DIR = "global-model-processed-data"
TRAIN_RATIO = 0.8
RANDOM_SEED = 42

random.seed(RANDOM_SEED)
```

```python
def ensure_dir(path):
    os.makedirs(path, exist_ok=True)

def reset_dir(path):
    if os.path.exists(path):
        shutil.rmtree(path)

def copy_files(files, target_dir):
    ensure_dir(target_dir)
    for f in files:
        machine_id = os.path.basename(os.path.dirname(os.path.dirname(f)))
        new_name = f"{machine_id}_{os.path.basename(f)}"
        shutil.copy2(f, os.path.join(target_dir, new_name))

def count_files(folder_path):
    return sum(len(files) for _, _, files in os.walk(folder_path))
```

```python
normal_files = []
abnormal_files = []

for machine_id in os.listdir(SOURCE_DIR):
    id_path = os.path.join(SOURCE_DIR, machine_id)
    if not os.path.isdir(id_path):
        continue

    normal_files.extend(
        glob(os.path.join(id_path, "normal", "*.wav"))
    )
    abnormal_files.extend(
        glob(os.path.join(id_path, "abnormal", "*.wav"))
    )

print(f"Found {len(normal_files)} normal files")
print(f"Found {len(abnormal_files)} abnormal files")
```

```python
random.shuffle(normal_files)

split_idx = int(TRAIN_RATIO * len(normal_files))
train_normal = normal_files[:split_idx]
test_normal = normal_files[split_idx:]
test_abnormal = abnormal_files
```

```python
copy_files(train_normal, os.path.join(DEST_DIR, "train", "normal"))
copy_files(test_normal, os.path.join(DEST_DIR, "test", "normal"))
copy_files(test_abnormal, os.path.join(DEST_DIR, "test", "abnormal"))
```

```python
print("\nTrain–Test split completed")
print(f"Train normal : {len(train_normal)}")
print(f"Test normal  : {len(test_normal)}")
print(f"Test abnormal: {len(test_abnormal)}")

print("\nVerified on disk:")
print("Train normal :", count_files("global-model-processed-data/train/normal"))
print("Test normal  :", count_files("global-model-processed-data/test/normal"))
print("Test abnormal:", count_files("global-model-processed-data/test/abnormal"))
```

```python
assert count_files("global-model-processed-data/train/normal") == len(train_normal)
assert count_files("global-model-processed-data/test/normal") == len(test_normal)
assert count_files("global-model-processed-data/test/abnormal") == len(test_abnormal)

print("\nDataset integrity verified")
```

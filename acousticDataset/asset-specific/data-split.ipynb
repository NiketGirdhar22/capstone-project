```python
import os
import shutil
import random
from glob import glob
```

```python
SOURCE_DIR = "../data/MIMII/fan"
DEST_DIR = "asset-specific-model-processed-data"
TRAIN_RATIO = 0.8
RANDOM_SEED = 42
```

```python
random.seed(RANDOM_SEED)

def ensure_dir(path):
    os.makedirs(path, exist_ok=True)

def copy_files(files, target_dir):
    ensure_dir(target_dir)
    for f in files:
        shutil.copy(f, target_dir)
```

```python
for machine_id in sorted(os.listdir(SOURCE_DIR)):
    id_path = os.path.join(SOURCE_DIR, machine_id)
    if not os.path.isdir(id_path):
        continue

    print(f"\nProcessing {machine_id}")

    normal_files = glob(os.path.join(id_path, "normal", "*.wav"))
    abnormal_files = glob(os.path.join(id_path, "abnormal", "*.wav"))

    random.shuffle(normal_files)

    split_idx = int(TRAIN_RATIO * len(normal_files))
    train_normal = normal_files[:split_idx]
    test_normal = normal_files[split_idx:]

    copy_files(
        train_normal,
        os.path.join(DEST_DIR, machine_id, "train", "normal")
    )
    copy_files(
        test_normal,
        os.path.join(DEST_DIR, machine_id, "test", "normal")
    )
    copy_files(
        abnormal_files,
        os.path.join(DEST_DIR, machine_id, "test", "abnormal")
    )

    print(f"  Train normal : {len(train_normal)}")
    print(f"  Test normal  : {len(test_normal)}")
    print(f"  Test abnormal: {len(abnormal_files)}")

print("\n ID-wise train–test split completed")
```

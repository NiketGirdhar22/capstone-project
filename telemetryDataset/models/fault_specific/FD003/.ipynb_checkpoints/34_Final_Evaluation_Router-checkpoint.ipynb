### LOADING AND CONFIG

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from tensorflow.keras.models import load_model
import os
import joblib

DATA_PATH = r'D:\PJT2\CMAPSSData' 
BASE_PATH = r'D:\PJT2\telemetryDataset'
PROCESSED_PATH = os.path.join(BASE_PATH, 'processed_data')
MODEL_PATH = r'./model_checkpoints'
```

```python
SCALER_PATH = os.path.join(PROCESSED_PATH, 'fd003_specialist_scaler.pkl')
ROUTER_PATH = os.path.join(PROCESSED_PATH, 'fd003_router.pkl')
CLUSTER_PATH = os.path.join(PROCESSED_PATH, 'fd003_cluster_scaler.pkl')
VALIDIDX = os.path.join(PROCESSED_PATH, 'fd003_valid_idx.pkl')

scaler_ts = joblib.load(SCALER_PATH)
router = joblib.load(ROUTER_PATH)
cluster_scaler = joblib.load(CLUSTER_PATH)
valid_idx = joblib.load(VALIDIDX)

print("Loaded frozen scaler and router.")
```

### ATTENTION LAYER 

```python
import tensorflow as tf
from tensorflow.keras.layers import Layer
class Attention(Layer):
    def __init__(self, **kwargs):
        super(Attention, self).__init__(**kwargs)
    def build(self, input_shape):
        self.W = self.add_weight(name='attention_weight', shape=(input_shape[-1], 1), initializer='random_normal', trainable=True)
        self.b = self.add_weight(name='attention_bias', shape=(input_shape[1], 1), initializer='zeros', trainable=True)
        super(Attention, self).build(input_shape)
    def call(self, x):
        e = tf.keras.backend.tanh(tf.keras.backend.dot(x, self.W) + self.b)
        a = tf.keras.backend.softmax(e, axis=1)
        output = x * a
        return tf.keras.backend.sum(output, axis=1)
```

### LOADING MODELS AND TEST DATA

```python
print("Loading Specialist Models...")
# Important: pass custom_objects so Keras knows what 'Attention' is
cols = ['unit_number', 'time_in_cycles', 'setting_1', 'setting_2', 'setting_3'] + \
       [f's{i}' for i in range(1, 22)]

model_c0 = load_model(os.path.join(MODEL_PATH, 'specialist_0_fd003.keras'), custom_objects={'Attention': Attention})
model_c1 = load_model(os.path.join(MODEL_PATH, 'specialist_1_fd003.keras'), custom_objects={'Attention': Attention})

test_df = pd.read_csv(os.path.join(DATA_PATH, 'test_FD003.txt'), sep='\s+', header=None, names=cols)
true_ruls = pd.read_csv(os.path.join(DATA_PATH, 'RUL_FD003.txt'), header=None, names=['RUL'])
```

### SETTING UP TEST FEATURES 

```python
# Load train just to recreate feature selection (no fitting!)
train_df = pd.read_csv(os.path.join(DATA_PATH, 'train_FD003.txt'),
                       sep='\s+', header=None, names=cols)

dataset_stds = train_df.describe().loc['std']
constant_features = dataset_stds[dataset_stds == 0].index.tolist()

drop_labels = constant_features + ['unit_number', 'time_in_cycles']
features = [c for c in train_df.columns if c not in drop_labels]
```

```python
test_df[features] = scaler_ts.transform(test_df[features])
test_df[features] = test_df[features].ewm(alpha=0.1).mean()
```

### ROUTING AND PREDICTING

```python
sensor_cols = [f's{i}' for i in range(1, 22)]
```

```python
print("\nRouting Test Engines and Predicting RUL...")
final_predictions = []
actuals = []
WINDOW_SIZE = 30

for unit_id in test_df['unit_number'].unique():
    unit_data = test_df[test_df['unit_number'] == unit_id]
    
    if len(unit_data) >= WINDOW_SIZE:
        # 1. Extract Test Signature
        test_sig = unit_data[sensor_cols].tail(5).mean().values
        test_sig_clean = test_sig[valid_idx].reshape(1, -1)
        test_sig_scaled = cluster_scaler.transform(test_sig_clean)
        
        # 2. Ask the Router
        assigned_cluster = router.predict(test_sig_scaled)[0]
        
        # 3. Extract the 30-cycle window
        seq = unit_data[features].values[-WINDOW_SIZE:].reshape(1, WINDOW_SIZE, len(features))
        
        # 4. Send to Specialist
        if assigned_cluster == 0:
            pred = model_c0.predict(seq, verbose=0)[0][0]
        else:
            pred = model_c1.predict(seq, verbose=0)[0][0]
            
        final_predictions.append(pred)
        actuals.append(true_ruls.iloc[unit_id-1].values[0])
```

### FINAL EVALUATION

```python
final_predictions = np.array(final_predictions)
actuals = np.array(actuals)

final_rmse = np.sqrt(np.mean((final_predictions - actuals) ** 2))

print("\n" + "="*50)
print(f"GLOBAL MODEL RMSE (Baseline): 21.04")
print(f"MIXTURE OF EXPERTS RMSE (Final): {final_rmse:.4f}")
print("="*50)

if final_rmse < 21.04:
    print("SUCCESS: The Specialists defeated the Global Model!")
```

```python
### VISUALISING
```

```python
results_df = pd.DataFrame({'True': actuals, 'Pred': final_predictions}).sort_values('True', ascending=False).reset_index(drop=True)
plt.figure(figsize=(15, 6))
plt.plot(results_df['True'], label='Actual RUL', color='black', linestyle='--')
plt.plot(results_df['Pred'], label='MoE Predicted RUL', color='green', alpha=0.8)
plt.title(f'FD003: Mixture of Experts Final Performance (RMSE: {final_rmse:.2f})')
plt.legend()
plt.show()
```

```python
### SAVING
```

```python
results_save_path = r'./fd003_moe_final_predictions.csv'

# We already have results_df from the plotting step above!
results_df.to_csv(results_save_path, index=False)

print(f"Final predictions successfully saved to: {os.path.abspath(results_save_path)}")
```

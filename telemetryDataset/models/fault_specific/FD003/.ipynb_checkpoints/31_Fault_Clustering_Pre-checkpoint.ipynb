### LOADING AND CONFIGURING

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
import os
import joblib

DATA_PATH = r'D:\PJT2\CMAPSSData' 
BASE_PATH = r'D:\PJT2\telemetryDataset'
PROCESSED_PATH = os.path.join(BASE_PATH, 'processed_data')
RUL_CAP = 125 
WINDOW_SIZE = 30 

cols = ['unit_number', 'time_in_cycles', 'setting_1', 'setting_2', 'setting_3'] + [f's{i}' for i in range(1, 22)]

print("Loading FD003 Raw Data for Fault Isolation...")
train_df = pd.read_csv(os.path.join(DATA_PATH, 'train_FD003.txt'), sep='\s+', header=None, names=cols)
test_df = pd.read_csv(os.path.join(DATA_PATH, 'test_FD003.txt'), sep='\s+', header=None, names=cols)
```

### FAULT ISOLATION

```python
print("\nExtracting Death Signatures..")
signatures = []
engine_ids = train_df['unit_number'].unique()
sensor_cols = [f's{i}' for i in range(1, 22)]

# Get the last 5 cycles of every engine to see HOW it broke
for unit in engine_ids:
    engine_data = train_df[train_df['unit_number'] == unit]
    end_of_life = engine_data[sensor_cols].tail(5).mean().values
    signatures.append(end_of_life)

signatures = np.array(signatures)
```

### CLEANING AND SCALING

```python
stds = signatures.std(axis=0)
valid_idx = np.where(stds > 0)[0] # Drop constant sensors for clustering
signatures_clean = signatures[:, valid_idx]

joblib.dump(valid_idx, os.path.join(PROCESSED_PATH, 'fd003_valid_idx.pkl'))

scaler_cluster = StandardScaler()
signatures_scaled = scaler_cluster.fit_transform(signatures_clean)
joblib.dump(scaler_cluster, os.path.join(PROCESSED_PATH, 'fd003_cluster_scaler.pkl'))
```

### K-MEANS, PLOTTING SEPARATION

```python
print("Running K-Means (k=2) to separate faults...")
kmeans = KMeans(n_clusters=2, random_state=42, n_init=20)
fault_labels = kmeans.fit_predict(signatures_scaled)

##Freezing router
ROUTER_PATH = os.path.join(PROCESSED_PATH, 'fd003_router.pkl')
os.makedirs(PROCESSED_PATH, exist_ok=True)
joblib.dump(kmeans, ROUTER_PATH)
print(f"Router saved to {ROUTER_PATH}")

fault_mapping = pd.DataFrame({'unit_number': engine_ids, 'fault_mode': fault_labels})
print("\nEngines assigned to Fault Modes:")
print(fault_mapping['fault_mode'].value_counts())

# Plot the separation
pca = PCA(n_components=2)
signatures_pca = pca.fit_transform(signatures_scaled)
pca_df = pd.DataFrame(signatures_pca, columns=['PCA 1', 'PCA 2'])
pca_df['Fault Mode'] = ['Cluster 0 (Fault A)' if x == 0 else 'Cluster 1 (Fault B)' for x in fault_labels]

plt.figure(figsize=(8, 5))
sns.scatterplot(data=pca_df, x='PCA 1', y='PCA 2', hue='Fault Mode', palette=['#FF9999', '#66B2FF'], s=100, edgecolor='black')
plt.title('Unsupervised Discovery of Two Distinct Fault Modes (FD003)')
plt.show()

print(os.path.abspath(PROCESSED_PATH))
```

## SPLITTING DATASETS

### RUL CALCULATION

```python
def add_rul(df):
    max_cycle = df.groupby('unit_number')['time_in_cycles'].max().reset_index()
    max_cycle.columns = ['unit_number', 'max_cycle']
    df = df.merge(max_cycle, on='unit_number', how='left')
    df['RUL'] = df['max_cycle'] - df['time_in_cycles']
    df = df.drop(columns=['max_cycle'])
    df['RUL'] = df['RUL'].clip(upper=RUL_CAP)
    return df

train_df = add_rul(train_df)
```

### MERGING, SCALING and SMOOTHING

```python
# Merge the fault labels back into the main dataframe
train_df = train_df.merge(fault_mapping, on='unit_number', how='left')

# Drop constants (from the whole dataset this time)
dataset_stds = train_df.describe().loc['std']
constant_features = dataset_stds[dataset_stds == 0].index.tolist()
drop_labels = constant_features + ['unit_number', 'time_in_cycles', 'RUL', 'fault_mode']
features = [c for c in train_df.columns if c not in drop_labels]

# Scale and Smooth
scaler = MinMaxScaler()
train_df[features] = scaler.fit_transform(train_df[features])
train_df[features] = train_df[features].ewm(alpha=0.1).mean()

# saving the scaler
os.makedirs(PROCESSED_PATH, exist_ok=True)

SCALER_PATH = os.path.join(PROCESSED_PATH, 'fd003_specialist_scaler.pkl')
joblib.dump(scaler, SCALER_PATH)

print(f"Scaler saved to {SCALER_PATH}")

# Sequence Generator
def gen_sequence(id_df, seq_length, seq_cols):
    data_matrix = id_df[seq_cols].values
    num_elements = data_matrix.shape[0]
    for start, stop in zip(range(0, num_elements-seq_length), range(seq_length, num_elements)):
        yield data_matrix[start:stop, :]

def gen_labels(id_df, seq_length, label):
    data_matrix = id_df[label].values
    return data_matrix[seq_length:]
```

### BUILDING SEPARATE ARRAYS

```python
X_train_c0, y_train_c0 = [], []
X_train_c1, y_train_c1 = [], []

for unit_id in train_df['unit_number'].unique():
    unit_data = train_df[train_df['unit_number'] == unit_id]
    fault_type = unit_data['fault_mode'].iloc[0] # Check which cluster this engine belongs to
    
    for seq in gen_sequence(unit_data, WINDOW_SIZE, features):
        if fault_type == 0: X_train_c0.append(seq)
        else: X_train_c1.append(seq)
            
    for label in gen_labels(unit_data, WINDOW_SIZE, 'RUL'):
        if fault_type == 0: y_train_c0.append(label)
        else: y_train_c1.append(label)

print(f"Cluster 0 Training Shape: {np.array(X_train_c0).shape}")
print(f"Cluster 1 Training Shape: {np.array(X_train_c1).shape}")
```

### SAVING

```python
import os
PROCESSED_PATH = r'../../../processed_data'

if not os.path.exists(PROCESSED_PATH):
    os.makedirs(PROCESSED_PATH)
    print(f"Created directory: {os.path.abspath(PROCESSED_PATH)}")

print("Saving arrays to corrected path...")
np.save(os.path.join(PROCESSED_PATH, 'X_train_FD003_c0.npy'), np.array(X_train_c0))
np.save(os.path.join(PROCESSED_PATH, 'y_train_FD003_c0.npy'), np.array(y_train_c0))
np.save(os.path.join(PROCESSED_PATH, 'X_train_FD003_c1.npy'), np.array(X_train_c1))
np.save(os.path.join(PROCESSED_PATH, 'y_train_FD003_c1.npy'), np.array(y_train_c1))

print("Success! Files saved.")
```
